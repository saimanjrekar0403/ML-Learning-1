# **Load Data**

```python
import pandas as pd

df = pd.read_csv("Dataset/Diabetes_data_og.csv")
df = df.drop_duplicates()
df = df.replace({"Diabetes_012": {2:1}})
df
```

# **Data preparation**

## Data seperation as X and Y

```python
y = df['Diabetes_012']
y
```

```python
x = df.drop(columns=['Diabetes_012'])
x
```

## Data splitting

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=100, stratify=y)
```

# **Model Building**

## Linear Regression

### **Training the model**

```python
from sklearn.linear_model import LinearRegression # type: ignore

lr = LinearRegression()
lr.fit(x_train, y_train)
```

### **Applying the model to make prediction**

```python
y_lr_train_pred = lr.predict(x_train)
y_lr_test_pred = lr.predict(x_test)
```

```python
y_lr_train_pred
```

```python
y_lr_test_pred
```

### **Model Performance**

```python
from sklearn.metrics import mean_squared_error, r2_score

lr_train_mse = mean_squared_error(y_train, y_lr_train_pred)
lr_train_r2 = r2_score(y_train, y_lr_train_pred)

lr_test_mse = mean_squared_error(y_test, y_lr_test_pred)
lr_test_r2 = r2_score(y_test, y_lr_test_pred)
```

```python
lr_results = pd.DataFrame({'Model': ['Linear Regression'], 
                           'Train MSE': [lr_train_mse], 
                           'Train R2': [lr_train_r2], 
                           'Test MSE': [lr_test_mse], 
                           'Test R2': [lr_test_r2]})
lr_results
```

## **Random Forest**

### **Training the model**

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(max_depth=10,
                            n_estimators=150,
                            random_state=100,
                            class_weight='balanced',
                            n_jobs=-1)
rf.fit(x_train, y_train)
```

### **Applying the model to make predictions**

```python
y_rf_train_pred = rf.predict(x_train)
y_rf_test_pred = rf.predict(x_test)
```

```python
y_rf_train_pred
```

```python
y_rf_test_pred
```

### **Model Performance**

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, y_rf_test_pred))
print(classification_report(y_train, y_rf_train_pred))
```

```python
from sklearn.metrics import mean_squared_error, r2_score

rf_train_mse = mean_squared_error(y_train, y_rf_train_pred)
rf_train_r2 = r2_score(y_train, y_rf_train_pred)

rf_test_mse = mean_squared_error(y_test, y_rf_test_pred)
rf_test_r2 = r2_score(y_test, y_rf_test_pred)
```

```python
rf_results = pd.DataFrame({'Model': ['Random Forest'],
                           'Train MSE': [rf_train_mse],
                           'Train R2': [rf_train_r2],
                           'Test MSE': [rf_test_mse],
                           'Test R2': [rf_test_r2]})
rf_results
```

## **Isolation Forest**

### **Training the model**

```python
from sklearn.ensemble import IsolationForest
import numpy as np

y_train_bin = (y_train > 0).astype(int)
y_test_bin = (y_test > 0).astype(int)

iso = IsolationForest(n_estimators=200,
                      contamination=y_train_bin.mean(),
                      random_state=100,
                      n_jobs=-1)
iso.fit(x_train[y_train == 0])
```

### **Applying the model to make predictions**

```python
# sklearn returns 1 = normal, -1 = anomaly, so flip to 0/1
y_iso_train_pred = (iso.predict(x_train) == -1).astype(int)
y_iso_test_pred = (iso.predict(x_test) == -1).astype(int)

# Higher score = more anomalous
y_iso_train_score = -iso.score_samples(x_train)
y_iso_test_score = -iso.score_samples(x_test)
```

```python
y_iso_train_pred
```

```python
y_iso_test_pred
```

### **Model Performance**

```python
from sklearn.metrics import roc_auc_score, recall_score, f1_score, accuracy_score

iso_train_auc = roc_auc_score(y_train_bin, y_iso_train_score)
iso_train_recall = recall_score(y_train_bin, y_iso_train_pred)
iso_train_f1 = f1_score(y_train_bin, y_iso_train_pred, average='macro')

iso_test_auc = roc_auc_score(y_test_bin, y_iso_test_score)
iso_test_recall = recall_score(y_test_bin, y_iso_test_pred)
iso_test_f1 = f1_score(y_test_bin, y_iso_test_pred, average='macro')
```

```python
iso_results = pd.DataFrame({'Model': ['Isolation Forest'],
                            'Train ROC-AUC': [iso_train_auc],
                            'Train Recall': [iso_train_recall],
                            'Train Macro F1': [iso_train_f1],
                            'Test ROC-AUC': [iso_test_auc],
                            'Test Recall': [iso_test_recall],
                            'Test Macro F1': [iso_test_f1]})
iso_results
```

```python
from sklearn.metrics import accuracy_score, f1_score, recall_score
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, HistGradientBoostingClassifier
import pandas as pd

models = {
    "Logistic Regression": LogisticRegression(max_iter=1000, class_weight='balanced'),
    "Random Forest": RandomForestClassifier(n_estimators=100, max_depth=10, class_weight='balanced', random_state=100, n_jobs=-1),
    "Hist Gradient Boosting": HistGradientBoostingClassifier(class_weight='balanced', random_state=100),
}

rows = []
for name, m in models.items():
    m.fit(x_train, y_train)
    pred = m.predict(x_test)
    rows.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, pred),
        "Macro F1": f1_score(y_test, pred, average='macro'),
        "Recall (class 1)": recall_score(y_test, pred, labels=[1], average=None)[0],
        "Recall (class 2)": recall_score(y_test, pred, labels=[2], average=None)[0],
    })

pd.DataFrame(rows)
```
