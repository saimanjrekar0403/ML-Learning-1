```python
import pandas as pd
df = pd.read_csv("Dataset/Diabetes_data_og.csv")
df = df.drop_duplicates()
df
```

```python
df["Diabetes_012"] = df["Diabetes_012"].replace(2.0, 1.0)
df
```

# **Data preparation**

## Data seperation as X and Y

```python
y = df['Diabetes_012']
y
```

```python
x = df.drop(columns=['Diabetes_012'])
x
```

## Data splitting

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=100, stratify=y)
```

# **Models**

```python
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.metrics import classification_report
HGBC = HistGradientBoostingClassifier(class_weight="balanced", random_state=42)
HGBC.fit(x_train, y_train)
print(classification_report(y_test, HGBC.predict(x_test)))
print(classification_report(y_train, HGBC.predict(x_train)))
```

```python
from lightgbm import LGBMClassifier
from sklearn.metrics import classification_report
LGBM = LGBMClassifier(class_weight="balanced", random_state=42)
LGBM.fit(x_train, y_train)
print(classification_report(y_test, LGBM.predict(x_test)))
print(classification_report(y_train, LGBM.predict(x_train)))
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
LGBM = LogisticRegression(class_weight="balanced", random_state=4)
LGBM.fit(x_train, y_train)
print(classification_report(y_test, LGBM.predict(x_test)))
print(classification_report(y_train, LGBM.predict(x_train)))
```
